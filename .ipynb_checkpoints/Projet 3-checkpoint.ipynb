## 1. Contexte et Définition du Problème


Dans le cadre de l'appel à manifestation d'intérêt de l'ADEME, notre structure CesiCDP a mené une étude approfondie sur la logistique du transport urbain. L'optimisation des tournées de livraison représente un enjeu écologique et économique majeur (réduction des émissions de GES, optimisation de la flotte).

Afin de proposer une solution réaliste et applicable aux contraintes des professionnels de la logistique, nous avons choisi de traiter le problème de Niveau A : le CVRPTW (Capacitated Vehicle Routing Problem with Time Windows).

Ce problème stipule qu'une flotte de $K$ camions de capacité identique $Q$ doit livrer un ensemble de clients depuis un dépôt central. Chaque client possède :
1. Une demande spécifique (poids/volume).
2. Une fenêtre de temps précise $[e_i, l_i]$ durant laquelle la livraison doit obligatoirement débuter.
3. Un temps de service sur place.

Les camions ont la possibilité d'attendre sur place si la fenêtre temporelle n'est pas encore ouverte, mais l'arrivée en retard est strictement interdite. L'objectif est de minimiser la distance totale et le temps total parcourue par la flotte.

## 2. Étude Théorique et Complexité Asymptotique

Avant de concevoir notre algorithme, il est fondamental d'étudier la complexité théorique du problème pour justifier nos choix techniques.

Théorème : Le CVRPTW (dans sa version d'optimisation) est NP-Difficile.

Preuve par réduction polynomiale :
Le Problème du Voyageur de Commerce (TSP) est un problème célèbre, prouvé NP-Complet par Richard Karp en 1972. Le TSP cherche le cycle hamiltonien le plus court passant par un ensemble de villes.

Considérons une instance de notre problème CVRPTW. Appliquons-y les restrictions suivantes :
* La flotte est réduite à un seul véhicule ($K = 1$).
* La capacité du véhicule est infinie ($Q = \infty$).
* Les fenêtres de temps sont totalement relâchées ($e_i = 0$ et $l_i = \infty$ pour tout client $i$).
* Le temps de service est nul.

Sous ces conditions, notre CVRPTW se réduit très exactement au problème du TSP. Puisque le TSP est un cas particulier du CVRPTW, le CVRPTW est au moins aussi difficile que le TSP. 
Par conséquent, le CVRPTW est NP-Difficile. Il est donc mathématiquement impossible de trouver un algorithme exact capable de le résoudre en temps polynomial (classe P). Cette démonstration justifie pleinement le recours à une métaheuristique pour trouver d'excellentes solutions en un temps de calcul acceptable.

## 3. Modélisation Mathématique (PLNE)

Pour formaliser ce problème, nous proposons un Programme Linéaire en Nombres Entiers (PLNE) « à deux indices » (formulation de flot sur les arcs), classique pour le VRPTW [Toth & Vigo, 2014].

### 3.1. Paramètres
* $N = \{1, ..., n\}$ : L'ensemble des clients.
* $V = \{0\} \cup N \cup \{n+1\}$ : L'ensemble des nœuds, où $0$ est le dépôt de départ et $n+1$ le dépôt d'arrivée (le même lieu physique).
* $A = \{(i,j) \in V^2 : i \neq j,\ i \neq n+1,\ j \neq 0\}$ : les arcs utilisables (on ne revient jamais vers le dépôt de départ, on ne repart jamais du dépôt d'arrivée).
* $K$ : Nombre de camions.
* $Q$ : Capacité maximale d'un camion.
* $q_i$ : Demande du client $i$ ($q_0 = q_{n+1} = 0$).
* $c_{ij}$ et $t_{ij}$ : Coût (distance) et temps de trajet de l'arc $(i, j)$ (sur les instances de Solomon, $c_{ij} = t_{ij}$ = distance euclidienne).
* $[e_i, l_i]$ : Fenêtre de temps du nœud $i$ ($[e_0, l_0]$ = horaires d'ouverture du dépôt).
* $s_i$ : Temps de service au nœud $i$ ($s_0 = 0$).
* $M_{ij} = \max(0,\ l_i + s_i + t_{ij} - e_j)$ : constante « Big-M » propre à chaque arc, la plus petite valeur qui désactive la contrainte temporelle quand l'arc n'est pas emprunté (un $M$ trop grand affaiblit la relaxation linéaire).

### 3.2. Variables de Décision
* $x_{ij} \in \{0, 1\}$ : Vaut $1$ si un véhicule emprunte l'arc $(i,j) \in A$, $0$ sinon.
* $w_i \ge 0$ : Heure de début de service au nœud $i$.
* $u_i \ge 0$ : Charge cumulée dans le camion après avoir visité $i$.

### 3.3. Fonction Objectif
Minimiser la distance totale de la flotte :
$$ \text{Min} \sum_{(i,j) \in A} c_{ij} x_{ij} $$

> Dans la littérature sur les instances de Solomon, l'objectif est souvent hiérarchique (d'abord le nombre de véhicules, puis la distance). Nous retenons l'objectif « distance seule », qui est celui des meilleures solutions connues (BKS) publiées par CVRPLIB et reprises par `vrplib` / PyVRP : c'est donc celui qui permet un calcul de gap honnête.

### 3.4. Contraintes
**Routage :**
1. Chaque client est visité exactement une fois : $\sum_{j : (i,j) \in A} x_{ij} = 1 \quad \forall i \in N$
2. Conservation du flux : $\sum_{i : (i,h) \in A} x_{ih} - \sum_{j : (h,j) \in A} x_{hj} = 0 \quad \forall h \in N$
3. Au plus $K$ camions partent du dépôt : $\sum_{j \in N} x_{0j} \le K$ (avec (2), autant de camions reviennent en $n+1$).

**Capacité (Élimination des sous-tours de Miller-Tucker-Zemlin) :**

4. Suivi du chargement : $u_i + q_j \le u_j + Q(1 - x_{ij}) \quad \forall (i,j) \in A,\ i, j \in N$
5. Respect de la capacité : $q_i \le u_i \le Q \quad \forall i \in N$

**Fenêtres de Temps :**

6. Continuité temporelle et temps d'attente : $w_i + s_i + t_{ij} \le w_j + M_{ij}(1 - x_{ij}) \quad \forall (i,j) \in A$
7. Respect strict de la fenêtre : $e_i \le w_i \le l_i \quad \forall i \in V$

**Domaines :** $x_{ij} \in \{0,1\}$, $w_i \ge 0$, $u_i \ge 0$.

**Remarques :**
* L'attente est autorisée car (6) est une inégalité : on peut avoir $w_j >$ heure d'arrivée, à condition que $w_j \ge e_j$ (7). L'arrivée en retard est interdite par $w_j \le l_j$.
* Les contraintes (4) et (6) éliminent chacune les sous-tours (charge ou heure strictement croissante le long d'un chemin).
* Prétraitement : tout arc tel que $e_i + s_i + t_{ij} > l_j$ ne peut jamais être emprunté, on fixe $x_{ij} = 0$ (on le retire de $A$). Sur les instances à fenêtres serrées, cela supprime une grande partie des arcs.
* Corrections par rapport à la première version : ensemble d'arcs $A$ explicite, Big-M calibré par arc ($M_{ij}$ pour le temps, $Q$ pour la charge), contrainte (1) indexée sur les arcs existants.

### 3.5. Utilisation de l'approche exacte
Le PLNE est implémenté ci-dessous et résolu avec le solveur **HiGHS** (inclus dans `scipy.optimize.milp`, aucune installation supplémentaire). Le problème étant NP-difficile, on ne l'utilise pas pour résoudre les instances complètes : il sert de **référence exacte** (optimum prouvé ou borne inférieure) sur des sous-instances de petite taille, afin de valider la métaheuristique (section 6.1).

## 4. Choix de la Métaheuristique : ALNS avec acceptation par Recuit Simulé

Puisque le problème est NP-difficile, nous utilisons une métaheuristique. Une première version reposait sur un Recuit Simulé « pur » (mouvements *relocate* / *swap* et pénalités). Elle produisait des solutions éloignées de l'optimum : un seul client déplacé à la fois ne suffit pas pour réorganiser des tournées très contraintes par les fenêtres de temps. Nous avons donc gardé le **Recuit Simulé comme critère d'acceptation**, en l'appliquant à de **gros voisinages** : c'est l'**ALNS** (*Adaptive Large Neighbourhood Search*) de Ropke & Pisinger (2006), l'une des méthodes de référence pour le VRPTW.

### 4.1. Définition d'une solution
Une solution est un ensemble de tournées $\{r_1, \dots, r_m\}$, $m \le K$, chaque tournée étant une séquence $0 \to c_1 \to \dots \to c_p \to 0$. **Toutes les solutions manipulées sont réalisables** : capacité et fenêtres de temps sont respectées à chaque instant. L'attente sur place est autorisée, le retard ne l'est pas. Le coût est la distance totale.

### 4.2. Principe : détruire puis reconstruire
À chaque itération :
1. **Destruction** : on retire $q$ clients de la solution courante ($q$ tiré dans $[q_{min}, q_{max}]$).
2. **Réparation** : on les réinsère un par un, à la meilleure position réalisable.
3. **Acceptation** : la nouvelle solution remplace la solution courante selon le critère du Recuit Simulé.

**Opérateurs de destruction** (voisinages) :

| Opérateur | Idée | Référence |
|---|---|---|
| `random` | $q$ clients au hasard (diversification) | Ropke & Pisinger 2006 |
| `worst` | clients dont la suppression fait gagner le plus de distance (tirage biaisé $y^3$) | Ropke & Pisinger 2006 |
| `shaw` | clients « liés » (proches en distance, en horaire et en demande), faciles à échanger entre eux | Shaw 1998 |
| `route` | vide entièrement une ou plusieurs petites tournées (réduction de flotte) | Pisinger & Ropke 2007 |
| `string` | retire des **chaînes** de clients consécutifs dans des tournées voisines | Christiaens & Vanden Berghe 2020 (SISR) |

**Opérateurs de réparation** :

| Opérateur | Idée |
|---|---|
| `greedy` | insère à chaque étape le client dont la meilleure insertion coûte le moins |
| `greedy_noise` | idem avec un bruit aléatoire sur les coûts (diversification) |
| `regret2`, `regret3` | insère d'abord le client qui « regrette » le plus de ne pas être inséré maintenant : $\sum_{h=2}^{k} (c_h - c_1)$, où $c_h$ est le coût de la $h$-ième meilleure tournée |

### 4.3. Mécanismes de guidage
* **Adaptation** : chaque opérateur $o$ a un poids $\omega_o$ et il est choisi par roulette. Tous les 100 itérations, $\omega_o \leftarrow (1-\rho)\,\omega_o + \rho\,\pi_o / n_o$, où $\pi_o$ cumule des points : 33 si nouvelle meilleure solution, 9 si amélioration de la solution courante, 13 si une solution moins bonne est acceptée ; $\rho = 0.1$.
* **Recuit Simulé** : une solution moins bonne de $\Delta$ est acceptée avec probabilité $P = \exp(-\Delta / T)$. $T_0$ est réglée pour qu'une solution 5 % plus chère que la solution initiale soit acceptée avec $P = 0.5$. Le refroidissement est géométrique, $T \leftarrow \alpha T$, jusqu'à $T_{fin} = 10^{-3} \, T_0$.
* **Critère d'arrêt** : un nombre d'itérations fixé (budget), pour des expériences reproductibles.

### 4.4. L'astuce qui rend l'algorithme rapide : test d'insertion en O(1)
Pour chaque tournée on maintient :
* $a_k$ : heure de début de service **au plus tôt** au $k$-ième nœud (calcul avant) : $a_k = \max(e_k,\ a_{k-1} + s_{k-1} + t_{k-1,k})$.
* $z_k$ : heure de début de service **au plus tard** qui garde la suite de la tournée réalisable (calcul arrière) : $z_k = \min(l_k,\ z_{k+1} - t_{k,k+1} - s_k)$.

Insérer $u$ entre $i$ (position $k$) et $j$ (position $k+1$) est réalisable **si et seulement si**
$$ a_u = \max(e_u,\ a_k + s_i + t_{iu}) \le l_u \quad \text{et} \quad a_u + s_u + t_{uj} \le z_{k+1}, $$
avec en plus la charge totale $\le Q$. Chaque test se fait donc en $O(1)$ au lieu de $O(n)$ (Savelsbergh 1992). Le cœur du code est compilé avec **Numba**, ce qui permet de faire des dizaines de milliers d'itérations en quelques secondes.

### 4.5. Complexité d'une itération
Avec $n$ clients, $m$ tournées et $q$ clients retirés : destruction en $O(n)$ (`random`, `worst`) ou $O(q \, n \log n)$ (`shaw`) ; réparation en $O(q \cdot n)$ pour le calcul initial des coûts d'insertion, puis $O(q \cdot (q \cdot m + |r|))$ grâce à un cache (seule la tournée modifiée est recalculée après chaque insertion).

## 5. Implémentation

### 5.0. Environnement et chargement des instances (VRPLIB)

**Correction de l'erreur `AttributeError: module 'vrplib' has no attribute 'download_instance'`** : depuis la version 2.0 de `vrplib`, les fonctions `download_instance` / `download_solution` ont été supprimées. L'ancien serveur CVRPLIB qu'elles interrogeaient n'existe plus. Il ne reste que `read_instance` / `read_solution`, qui lisent des fichiers locaux.

Nous téléchargeons donc une seule fois les instances (et leurs meilleures solutions connues) depuis le dépôt GitHub de référence [PyVRP/Instances](https://github.com/PyVRP/Instances), dans le dossier `instances/`, puis nous les lisons avec `vrplib.read_instance`.

⚠️ **Convention de distance** : les BKS de ce dépôt suivent la convention DIMACS, c'est-à-dire des distances **tronquées à une décimale** ($\lfloor 10\,d \rfloor / 10$). Nous l'appliquons aussi : sans elle, le coût optimal de C101 serait 828.94 au lieu de 827.3 et tous les gaps seraient faussés. Les instances CVRP « X » utilisent l'arrondi entier TSPLIB (`nint`).

```python
import math
import os
import time
import urllib.request

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import vrplib
from numba import njit
from scipy import stats
from scipy.optimize import milp, LinearConstraint, Bounds
from scipy.sparse import lil_matrix

INST_DIR = "instances"
RES_DIR = "results"
os.makedirs(RES_DIR, exist_ok=True)
GITHUB = "https://raw.githubusercontent.com/PyVRP/Instances/main"


def remote_folder(name, collection):
    """Dossier du dépôt PyVRP/Instances contenant l'instance."""
    if collection == "solomon":
        return "VRPTW/Solomon"
    if collection == "cvrp":
        return "CVRP"
    size = int(name.split("_")[1])          # Gehring & Homberger : C1_2_1 -> 200 clients
    return "VRPTW" if size == 10 else f"VRPTW/GH{size}00"


def load_instance(name, collection="solomon"):
    """Télécharge (si besoin) puis lit une instance et sa meilleure solution connue (BKS)."""
    folder = os.path.join(INST_DIR, collection)
    os.makedirs(folder, exist_ok=True)
    paths = {}
    for ext in ("vrp", "sol"):
        paths[ext] = os.path.join(folder, f"{name}.{ext}")
        if not os.path.exists(paths[ext]):
            urllib.request.urlretrieve(f"{GITHUB}/{remote_folder(name, collection)}/{name}.{ext}",
                                       paths[ext])
    return vrplib.read_instance(paths["vrp"]), vrplib.read_solution(paths["sol"])


def family(name):
    """C101 -> C1, RC2_2_1 -> RC2, X-n101-k25 -> X."""
    if name.startswith("X"):
        return "X"
    return name.split("_")[0] if "_" in name else name[:-2]


SOLOMON = ([f"C1{i:02d}" for i in range(1, 10)] + [f"C2{i:02d}" for i in range(1, 9)]
           + [f"R1{i:02d}" for i in range(1, 13)] + [f"R2{i:02d}" for i in range(1, 12)]
           + [f"RC1{i:02d}" for i in range(1, 9)] + [f"RC2{i:02d}" for i in range(1, 9)])
print(len(SOLOMON), "instances de Solomon")
```

### 5.1. Données
`build_data` transforme une instance `vrplib` en tableaux NumPy : matrice des distances (= temps de trajet), demandes, fenêtres $[e_i, l_i]$, temps de service, capacité $Q$ et flotte $K$. Pour un CVRP (sans fenêtres), les fenêtres sont relâchées à $[0, +\infty[$ : le même code résout donc les deux problèmes (niveaux B et A).

```python
INF = 1e18
BIG = 1e6  # coût "plafond" utilisé dans le calcul du regret

# Opérateurs
DESTROY_NAMES = ["random", "worst", "shaw", "route", "string"]
REPAIR_NAMES = ["greedy", "greedy_noise", "regret2", "regret3"]


def build_data(instance, rounding="dimacs"):
    """Construit les tableaux numpy utilisés par le solveur à partir d'une instance vrplib."""
    coords = instance["node_coord"].astype(np.float64)
    n_nodes = coords.shape[0]
    diff = coords[:, None, :] - coords[None, :, :]
    dist = np.sqrt((diff ** 2).sum(-1))
    if rounding == "dimacs":        # convention DIMACS : troncature à 1 décimale (Solomon / GH)
        dist = np.floor(dist * 10.0) / 10.0
    elif rounding == "nint":        # convention TSPLIB EUC_2D : arrondi à l'entier (CVRP X)
        dist = np.floor(dist + 0.5)

    demand = instance["demand"].astype(np.float64)
    if "time_window" in instance:
        tw = instance["time_window"].astype(np.float64)
        e, l = tw[:, 0].copy(), tw[:, 1].copy()
    else:                           # CVRP pur : fenêtres totalement relâchées
        e, l = np.zeros(n_nodes), np.full(n_nodes, 1e12)
    st = instance.get("service_time", 0)
    s = np.full(n_nodes, float(st)) if np.isscalar(st) else np.asarray(st, np.float64).copy()
    s[0] = 0.0                      # pas de temps de service au dépôt

    n_cust = n_nodes - 1
    k_max = int(instance.get("vehicles", n_cust))
    return dict(coords=coords, dist=dist, demand=demand, e=e, l=l, s=s,
                Q=float(instance["capacity"]), K=k_max, n=n_cust)
```

### 5.2. Primitives sur les tournées
* `update_route` recalcule en $O(|r|)$ la charge, le coût et les tableaux $a$ (au plus tôt) et $z$ (au plus tard) d'une tournée.
* `best_insert_route` cherche la meilleure position d'insertion d'un client dans une tournée, avec le test de faisabilité en $O(1)$ par position (section 4.4). Comme $a_k$ est croissant, on arrête le parcours dès que $a_k > l_u$.
* `touch` sauvegarde une tournée avant sa première modification : si la nouvelle solution est rejetée, seules les tournées modifiées sont restaurées (pas de copie complète de la solution à chaque itération).

```python
@njit(cache=True)
def update_route(r, R, L, A, Z, load, rcost, dist, demand, e, l, s):
    """Recalcule charge, coût, heures au plus tôt (A) et au plus tard (Z) de la tournée r."""
    m = L[r]
    ld = 0.0
    c = 0.0
    A[r, 0] = e[0]
    for k in range(1, m):
        i = R[r, k - 1]
        j = R[r, k]
        c += dist[i, j]
        ld += demand[j]
        t = A[r, k - 1] + s[i] + dist[i, j]
        A[r, k] = t if t > e[j] else e[j]          # attente si arrivée avant l'ouverture
    Z[r, m - 1] = l[0]
    for k in range(m - 2, -1, -1):
        i = R[r, k]
        j = R[r, k + 1]
        t = Z[r, k + 1] - dist[i, j] - s[i]
        Z[r, k] = t if t < l[i] else l[i]
    load[r] = ld
    rcost[r] = c


@njit(cache=True)
def best_insert_route(u, r, R, L, A, Z, load, Q, dist, demand, e, l, s):
    """Meilleure position (coût marginal minimal) pour insérer u dans r. O(|r|)."""
    if load[r] + demand[u] > Q:
        return INF, -1
    best = INF
    bp = -1
    for p in range(L[r] - 1):
        if A[r, p] > l[u]:          # A est croissant : plus aucune position possible ensuite
            break
        i = R[r, p]
        j = R[r, p + 1]
        arr = A[r, p] + s[i] + dist[i, u]
        if arr < e[u]:
            arr = e[u]
        if arr > l[u]:
            continue
        if arr + s[u] + dist[u, j] > Z[r, p + 1]:   # le reste de la tournée deviendrait en retard
            continue
        delta = dist[i, u] + dist[u, j] - dist[i, j]
        if delta < best:
            best = delta
            bp = p
    return best, bp


@njit(cache=True)
def touch(r, R, L, Rb, Lb, touched, tlist, nt):
    """Sauvegarde la tournée r avant sa première modification dans l'itération."""
    if not touched[r]:
        touched[r] = True
        Lb[r] = L[r]
        for k in range(L[r]):
            Rb[r, k] = R[r, k]
        tlist[nt] = r
        return nt + 1
    return nt


@njit(cache=True)
def insert_at(u, r, p, R, L, route_of):
    for k in range(L[r], p + 1, -1):
        R[r, k] = R[r, k - 1]
    R[r, p + 1] = u
    L[r] += 1
    route_of[u] = r


@njit(cache=True)
def remove_cust(c, R, L, route_of):
    r = route_of[c]
    k = 1
    while R[r, k] != c:
        k += 1
    for h in range(k, L[r] - 1):
        R[r, h] = R[r, h + 1]
    L[r] -= 1
    route_of[c] = -1
    return r
```

### 5.3. Opérateurs de réparation
Insertion gloutonne (avec ou sans bruit) et heuristique du regret-$k$. Les coûts d'insertion sont mis en cache dans une matrice (client × tournée). Après chaque insertion, seule la colonne de la tournée modifiée est recalculée. Une seule tournée vide est proposée à la fois, pour ne pas fausser le calcul du regret.

```python
@njit(cache=True)
def repair(U, nU, regret_k, noise, R, L, A, Z, load, rcost, route_of,
           Rb, Lb, touched, tlist, nt, Q, dist, demand, e, l, s, dmax):
    """Réinsère les clients U[:nU]. Retourne (succès, nt).
    regret_k = 1 -> glouton (meilleure insertion) ; k >= 2 -> heuristique du regret-k.
    Les coûts d'insertion sont mis en cache par (client, tournée) et seule la colonne de
    la tournée modifiée est recalculée après chaque insertion."""
    maxR = L.shape[0]
    C = np.full((nU, maxR), INF)
    P = np.full((nU, maxR), -1, np.int64)
    emp = -1                                # une seule tournée vide "candidate"
    for r in range(maxR):
        if L[r] == 2:
            emp = r
            break
    for r in range(maxR):
        if L[r] > 2 or r == emp:
            for i in range(nU):
                C[i, r], P[i, r] = best_insert_route(U[i], r, R, L, A, Z, load, Q,
                                                     dist, demand, e, l, s)
    alive = np.ones(nU, np.bool_)
    for _ in range(nU):
        best_score = -INF
        bi = -1
        br = -1
        for i in range(nU):
            if not alive[i]:
                continue
            # k meilleurs coûts sur les tournées (sélection partielle)
            c1 = INF
            r1 = -1
            for r in range(maxR):
                if C[i, r] < c1:
                    c1 = C[i, r]
                    r1 = r
            if r1 == -1:
                continue
            if regret_k == 1:
                c = c1
                if noise:
                    c = c1 + (2.0 * np.random.random() - 1.0) * 0.025 * dmax
                    if c < 0.0:
                        c = 0.0
                score = -c
            else:
                score = 0.0
                prev = c1
                prev_r = r1
                used = np.empty(regret_k, np.int64)
                used[0] = r1
                for h in range(1, regret_k):
                    ch = INF
                    rh = -1
                    for r in range(maxR):
                        if C[i, r] < ch and C[i, r] >= prev:
                            skip = False
                            for g in range(h):
                                if used[g] == r:
                                    skip = True
                                    break
                            if not skip:
                                ch = C[i, r]
                                rh = r
                    used[h] = rh
                    score += (ch if ch < BIG else BIG) - c1
                    prev = ch
                    prev_r = rh
                score = score * 1000.0 - c1      # départage par le coût d'insertion
            if score > best_score:
                best_score = score
                bi = i
                br = r1
        if bi == -1:
            return False, nt                    # client(s) impossible(s) à insérer
        nt = touch(br, R, L, Rb, Lb, touched, tlist, nt)
        insert_at(U[bi], br, P[bi, br], R, L, route_of)
        update_route(br, R, L, A, Z, load, rcost, dist, demand, e, l, s)
        alive[bi] = False
        for i in range(nU):
            if alive[i]:
                C[i, br], P[i, br] = best_insert_route(U[i], br, R, L, A, Z, load, Q,
                                                       dist, demand, e, l, s)
        if br == emp:                            # la tournée vide a été utilisée : on en ouvre une autre
            emp = -1
            for r in range(maxR):
                if L[r] == 2:
                    emp = r
                    break
            if emp >= 0:
                for i in range(nU):
                    if alive[i]:
                        C[i, emp], P[i, emp] = best_insert_route(U[i], emp, R, L, A, Z, load, Q,
                                                                 dist, demand, e, l, s)
    return True, nt
```

### 5.4. Opérateurs de destruction
`random`, `worst`, `shaw`, `route` et `string` (SISR), décrits en section 4.2. Les tirages biaisés $\lfloor y^p \cdot |L| \rfloor$ ($y$ uniforme sur $[0,1[$) favorisent le début d'une liste triée tout en gardant de l'aléa.

```python
@njit(cache=True)
def pick_power(nb, p):
    """Index aléatoire biaisé vers 0 (sélection randomisée de Ropke & Pisinger)."""
    return int((np.random.random() ** p) * nb)


@njit(cache=True)
def destroy(op, q, n, R, L, A, route_of, Rb, Lb, touched, tlist, nt,
            dist, demand, neigh, dmax, tmax, qmax):
    """Retire environ q clients selon l'opérateur op. Retourne (liste, nb retirés, nt)."""
    maxR = L.shape[0]
    rem = np.empty(n, np.int64)
    nrem = 0
    removed = np.zeros(n + 1, np.bool_)

    if op == 0:                                   # --- aléatoire
        perm = np.random.permutation(n) + 1
        for k in range(q):
            rem[nrem] = perm[k]
            nrem += 1

    elif op == 1:                                 # --- pire coût (gain de suppression)
        gain = np.zeros(n)
        for r in range(maxR):
            for k in range(1, L[r] - 1):
                i, c, j = R[r, k - 1], R[r, k], R[r, k + 1]
                gain[c - 1] = dist[i, c] + dist[c, j] - dist[i, j]
        order = np.argsort(-gain) + 1
        avail = order.copy()
        na = n
        for _ in range(q):
            idx = pick_power(na, 3.0)
            rem[nrem] = avail[idx]
            nrem += 1
            for h in range(idx, na - 1):
                avail[h] = avail[h + 1]
            na -= 1

    elif op == 2:                                 # --- Shaw (clients "liés")
        T = np.zeros(n + 1)
        for r in range(maxR):
            for k in range(1, L[r] - 1):
                T[R[r, k]] = A[r, k]
        first = np.random.randint(1, n + 1)
        rem[0] = first
        nrem = 1
        removed[first] = True
        while nrem < q:
            ref = rem[np.random.randint(0, nrem)]
            rel = np.full(n, INF)
            for c in range(1, n + 1):
                if not removed[c]:
                    rel[c - 1] = (9.0 * dist[ref, c] / dmax
                                  + 3.0 * abs(T[ref] - T[c]) / tmax
                                  + 2.0 * abs(demand[ref] - demand[c]) / qmax)
            order = np.argsort(rel)
            idx = pick_power(n - nrem, 6.0)
            c = order[idx] + 1
            rem[nrem] = c
            nrem += 1
            removed[c] = True

    elif op == 3:                                 # --- suppression de tournée(s) (petites en priorité)
        while nrem < max(1, q // 3):
            best_r = -1
            for _ in range(2):                    # tournoi de taille 2 sur les tournées non vides
                for _t in range(20):
                    r = np.random.randint(0, maxR)
                    if L[r] > 2 and not touched[r]:
                        break
                if L[r] > 2 and not touched[r]:
                    if best_r == -1 or L[r] < L[best_r]:
                        best_r = r
            if best_r == -1:
                break
            nt = touch(best_r, R, L, Rb, Lb, touched, tlist, nt)
            for k in range(1, L[best_r] - 1):
                rem[nrem] = R[best_r, k]
                nrem += 1
        for k in range(nrem):
            route_of[rem[k]] = -1
        for r in range(maxR):
            if touched[r]:
                L[r] = 2
                R[r, 1] = 0
        return rem, nrem, nt

    else:                                         # --- chaînes (SISR, Christiaens & Vanden Berghe 2020)
        nonempty = 0
        for r in range(maxR):
            if L[r] > 2:
                nonempty += 1
        avg_len = n / max(1, nonempty)
        ls_max = min(10.0, avg_len)
        ks_max = 4.0 * q / (1.0 + ls_max) - 1.0
        ks = int(np.random.random() * ks_max) + 1
        seed = np.random.randint(1, n + 1)
        ruined = np.zeros(maxR, np.bool_)
        nr = 0
        for h in range(-1, neigh.shape[1]):
            if nr >= ks:
                break
            c = seed if h == -1 else neigh[seed, h]
            if c == 0 or removed[c]:
                continue
            r = route_of[c]
            if r < 0 or ruined[r]:
                continue
            m = L[r] - 2
            lmax = min(float(m), ls_max)
            ln = int(np.random.random() * lmax) + 1
            pos = 1
            while R[r, pos] != c:
                pos += 1
            start = pos - np.random.randint(0, ln)           # chaîne de longueur ln contenant c
            if start < 1:
                start = 1
            if start + ln - 1 > m:
                start = m - ln + 1
            for k in range(start, start + ln):
                cc = R[r, k]
                rem[nrem] = cc
                nrem += 1
                removed[cc] = True
            ruined[r] = True
            nr += 1

    for k in range(nrem):
        c = rem[k]
        nt = touch(route_of[c], R, L, Rb, Lb, touched, tlist, nt)
        remove_cust(c, R, L, route_of)
    return rem, nrem, nt
```

### 5.5. Boucle principale ALNS
Étapes : (1) **initialisation** par insertion regret-2 de tous les clients à partir de tournées vides ; (2) **itérations** : choix des opérateurs par roulette, destruction, réparation, acceptation par recuit simulé, mise à jour des scores ; (3) **adaptation** des poids tous les `segment` itérations ; (4) **arrêt** après `n_iter` itérations. L'historique (coût courant / meilleur coût) est conservé pour tracer la convergence.

```python
@njit(cache=True)
def alns_core(dist, demand, e, l, s, Q, maxR, n_iter, seed, q_min, q_max,
              start_worse, end_temp_ratio, segment, reaction, neigh):
    np.random.seed(seed)
    n = dist.shape[0] - 1
    maxLen = n + 2
    R = np.zeros((maxR, maxLen), np.int64)
    L = np.full(maxR, 2, np.int64)
    A = np.zeros((maxR, maxLen))
    Z = np.zeros((maxR, maxLen))
    load = np.zeros(maxR)
    rcost = np.zeros(maxR)
    route_of = np.full(n + 1, -1, np.int64)
    Rb = np.zeros((maxR, maxLen), np.int64)
    Lb = np.zeros(maxR, np.int64)
    touched = np.zeros(maxR, np.bool_)
    tlist = np.zeros(maxR, np.int64)
    for r in range(maxR):
        update_route(r, R, L, A, Z, load, rcost, dist, demand, e, l, s)

    dmax = dist.max()
    tmax = l[0] - e[0]
    qmax = max(1.0, demand.max() - demand[1:].min())

    # --- solution initiale : insertion regret-2 de tous les clients
    U = np.arange(1, n + 1)
    ok, nt = repair(U, n, 2, False, R, L, A, Z, load, rcost, route_of,
                    Rb, Lb, touched, tlist, 0, Q, dist, demand, e, l, s, dmax)
    if not ok:
        ok, nt = repair(U[np.argsort(-(l[1:] - e[1:]))[::-1]], n, 1, False, R, L, A, Z, load,
                        rcost, route_of, Rb, Lb, touched, tlist, 0, Q, dist, demand, e, l, s, dmax)
    init_cost = rcost.sum()
    cur_cost = init_cost
    best_cost = init_cost
    bestR = R.copy()
    bestL = L.copy()

    nd, nr_ = 5, 4
    wd = np.ones(nd)
    wr = np.ones(nr_)
    sd = np.zeros(nd)
    sr = np.zeros(nr_)
    cd = np.zeros(nd)
    cr = np.zeros(nr_)
    used_d = np.zeros(nd, np.int64)       # statistiques globales d'utilisation
    best_d = np.zeros(nd, np.int64)

    T0 = -start_worse * init_cost / math.log(0.5)
    alpha = end_temp_ratio ** (1.0 / max(1, n_iter))
    T = T0
    hist_cur = np.zeros(n_iter)
    hist_best = np.zeros(n_iter)

    for it in range(n_iter):
        # choix des opérateurs par roulette sur les poids adaptatifs
        x = np.random.random() * wd.sum()
        d = 0
        while x > wd[d] and d < nd - 1:
            x -= wd[d]
            d += 1
        x = np.random.random() * wr.sum()
        rp = 0
        while x > wr[rp] and rp < nr_ - 1:
            x -= wr[rp]
            rp += 1

        nt = 0
        for k in range(maxR):
            touched[k] = False
        q = np.random.randint(q_min, q_max + 1)
        rem, nrem, nt = destroy(d, q, n, R, L, A, route_of, Rb, Lb, touched, tlist, nt,
                                dist, demand, neigh, dmax, tmax, qmax)
        for k in range(nt):
            update_route(tlist[k], R, L, A, Z, load, rcost, dist, demand, e, l, s)
        rk = 1 if rp <= 1 else rp
        ok, nt = repair(rem[:nrem].copy(), nrem, rk, rp == 1, R, L, A, Z, load, rcost, route_of,
                        Rb, Lb, touched, tlist, nt, Q, dist, demand, e, l, s, dmax)

        score = 0.0
        accepted = False
        if ok:
            new_cost = rcost.sum()
            if new_cost < best_cost - 1e-7:
                best_cost = new_cost
                bestR[:, :] = R
                bestL[:] = L
                score = 33.0
                accepted = True
                best_d[d] += 1
            elif new_cost < cur_cost - 1e-7:
                score = 9.0
                accepted = True
            elif np.random.random() < math.exp(-(new_cost - cur_cost) / T):
                score = 13.0 if new_cost > cur_cost + 1e-7 else 0.0
                accepted = True
            if accepted:
                cur_cost = new_cost
        if not accepted:                          # on restaure les tournées modifiées
            for k in range(nt):
                r = tlist[k]
                L[r] = Lb[r]
                for h in range(Lb[r]):
                    R[r, h] = Rb[r, h]
                for h in range(1, L[r] - 1):
                    route_of[R[r, h]] = r
                update_route(r, R, L, A, Z, load, rcost, dist, demand, e, l, s)

        used_d[d] += 1
        sd[d] += score
        sr[rp] += score
        cd[d] += 1
        cr[rp] += 1
        if (it + 1) % segment == 0:               # mise à jour des poids (réaction)
            for k in range(nd):
                if cd[k] > 0:
                    wd[k] = (1 - reaction) * wd[k] + reaction * sd[k] / cd[k]
                wd[k] = max(wd[k], 0.05)
                sd[k] = 0.0
                cd[k] = 0
            for k in range(nr_):
                if cr[k] > 0:
                    wr[k] = (1 - reaction) * wr[k] + reaction * sr[k] / cr[k]
                wr[k] = max(wr[k], 0.05)
                sr[k] = 0.0
                cr[k] = 0
        T *= alpha
        hist_cur[it] = cur_cost
        hist_best[it] = best_cost

    return bestR, bestL, best_cost, init_cost, hist_cur, hist_best, wd, wr, used_d, best_d
```

### 5.6. Interface Python et vérificateur indépendant
`solve` lance l'algorithme et renvoie les tournées. `check_solution` est un vérificateur écrit **séparément, en Python pur**. Il recalcule le coût et contrôle que chaque client est servi une seule fois, que la capacité, la flotte et les fenêtres de temps sont respectées. Chaque résultat présenté dans ce notebook passe par ce vérificateur.

```python
def solve(data, n_iter=20000, seed=0, q_min=None, q_max=None, start_worse=0.05,
          end_temp_ratio=1e-3, segment=100, reaction=0.1, n_neigh=50):
    """Lance l'ALNS et renvoie un dictionnaire de résultats (routes, coût, temps, historique)."""
    n = data["n"]
    q_min = q_min if q_min is not None else min(4, n)
    q_max = q_max if q_max is not None else max(q_min, min(int(0.3 * n), 60))
    dist = data["dist"]
    kn = min(n_neigh, n)
    neigh = np.argsort(dist, axis=1)[:, 1:kn + 1].astype(np.int64)
    t0 = time.perf_counter()
    out = alns_core(dist, data["demand"], data["e"], data["l"], data["s"], data["Q"],
                    data["K"], n_iter, seed, q_min, q_max, start_worse, end_temp_ratio,
                    segment, reaction, neigh)
    elapsed = time.perf_counter() - t0
    bestR, bestL, best_cost, init_cost, hc, hb, wd, wr, used_d, best_d = out
    routes = [list(bestR[r, 1:bestL[r] - 1]) for r in range(len(bestL)) if bestL[r] > 2]
    return dict(routes=routes, cost=best_cost, init_cost=init_cost, time=elapsed,
                hist_cur=hc, hist_best=hb, w_destroy=wd, w_repair=wr,
                used_destroy=used_d, best_destroy=best_d)


def check_solution(data, routes):
    """Vérification indépendante (Python pur) : visites, capacité, fenêtres de temps."""
    dist, dem, e, l, s, Q = (data[k] for k in ("dist", "demand", "e", "l", "s", "Q"))
    seen = sorted(c for r in routes for c in r)
    assert seen == list(range(1, data["n"] + 1)), "chaque client doit être visité une fois"
    assert len(routes) <= data["K"], "trop de véhicules"
    total = 0.0
    for r in routes:
        assert sum(dem[c] for c in r) <= Q + 1e-9, "capacité dépassée"
        t, prev = e[0], 0
        for c in list(r) + [0]:
            t = max(e[c], t + s[prev] + dist[prev, c])
            assert t <= l[c] + 1e-9, f"fenêtre violée au nœud {c}"
            total += dist[prev, c]
            prev = c
    return total
```

### 5.7. Cas de test
Tests unitaires sur une petite instance construite à la main, dont on connaît la solution optimale, puis tests négatifs : le vérificateur doit rejeter une solution qui viole une fenêtre ou la capacité.

```python
def toy_instance():
    """Dépôt au centre, 4 clients aux 4 coins d'un carré de côté 20. Q = 20, chaque demande = 10.
    Les fenêtres imposent l'ordre : client 1 tôt, client 3 tard."""
    return dict(node_coord=np.array([[0, 0], [10, 10], [-10, 10], [-10, -10], [10, -10]]),
                demand=np.array([0, 10, 10, 10, 10]), capacity=20, vehicles=4,
                time_window=np.array([[0, 1000], [0, 20], [0, 1000], [200, 300], [0, 1000]]),
                service_time=5)


toy = build_data(toy_instance())
res = solve(toy, n_iter=200, seed=0)
cost = check_solution(toy, res["routes"])
print("tournées :", res["routes"], "| coût :", cost)

# 1) capacité 20 et demandes de 10 => au moins 2 tournées de 2 clients
assert len(res["routes"]) == 2 and all(len(r) == 2 for r in res["routes"])
# 2) optimum = 2 tournées (dépôt -> coin -> coin adjacent -> dépôt) = 2 * (14.1 + 20 + 14.1)
assert abs(cost - 2 * (14.1 + 20 + 14.1)) < 1e-6
# 3) le client 3 ne peut pas être servi avant t = 200 : sa fenêtre est respectée
for neg_routes, msg in [([[3, 1], [2, 4]], "fenêtre"), ([[1, 2, 3], [4]], "capacité")]:
    try:
        check_solution(toy, neg_routes)
        raise RuntimeError("le vérificateur aurait dû rejeter cette solution")
    except AssertionError as err:
        print(f"solution invalide ({msg}) correctement rejetée :", err)
print("Tous les tests passent.")
```

### 5.8. Démonstration
Le premier appel compile le code Numba (environ 30 s la première fois, ensuite le cache `cache=True` le rend instantané). On résout ensuite R101 (100 clients, clients dispersés, fenêtres serrées) et RC201 (mi-groupés, fenêtres larges).

```python
def plot_solution(data, routes, title, ax):
    xy = data["coords"]
    cmap = plt.get_cmap("tab20")
    for k, r in enumerate(routes):
        path = [0] + list(r) + [0]
        ax.plot(xy[path, 0], xy[path, 1], "-", lw=1.3, color=cmap(k % 20))
    ax.scatter(xy[1:, 0], xy[1:, 1], s=10, c="k", zorder=3)
    ax.scatter(xy[0, 0], xy[0, 1], marker="s", s=90, c="red", zorder=4, label="dépôt")
    ax.set_title(title, fontsize=10)
    ax.set_aspect("equal")
    ax.legend(loc="upper right", fontsize=8)


t = time.perf_counter()
solve(toy, n_iter=10, seed=0)
print(f"compilation / chargement du cache Numba : {time.perf_counter() - t:.1f} s")

demo = {}
fig, axes = plt.subplots(2, 2, figsize=(13, 11))
for col, name in enumerate(["R101", "RC201"]):
    inst, bks = load_instance(name, "solomon")
    data = build_data(inst)
    res = solve(data, n_iter=10000, seed=1)
    cost = check_solution(data, res["routes"])
    gap = 100 * (cost - bks["cost"]) / bks["cost"]
    demo[name] = res
    print(f"{name}: coût = {cost:.1f} | BKS = {bks['cost']} | gap = {gap:.2f} % | "
          f"véhicules = {len(res['routes'])} (BKS : {len(bks['routes'])}) | "
          f"temps = {res['time']:.2f} s | solution initiale = {res['init_cost']:.1f}")
    plot_solution(data, res["routes"], f"{name} – ALNS : {cost:.1f} (gap {gap:.2f} %)", axes[0, col])
    ax = axes[1, col]
    ax.plot(res["hist_cur"], lw=0.6, alpha=0.7, label="solution courante")
    ax.plot(res["hist_best"], lw=1.8, label="meilleure solution")
    ax.axhline(bks["cost"], color="k", ls="--", lw=1, label="BKS")
    ax.set_xlabel("itération")
    ax.set_ylabel("distance totale")
    ax.set_title(f"{name} – convergence", fontsize=10)
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

### 6.0. Plan d'expérience

| # | Question | Instances | Facteur étudié | Répétitions |
|---|---|---|---|---|
| 6.1 | L'ALNS trouve-t-il l'optimum quand on le connaît ? | sous-instances de Solomon (10, 25, 50 clients) | méthode exacte vs heuristique | 1 |
| 6.2 | Qualité vs BKS (validation VRPLIB) | **56 instances de Solomon** (6 familles) | répartition géographique (C/R/RC) × horizon (1 = serré, 2 = large) | 3 graines |
| 6.3 | Passage à l'échelle | Gehring & Homberger 200 → 1000 clients + instances générées de 2000 et 3000 clients | taille $n$ | 1 graine |
| 6.4 | Convergence | R101, RC201 | budget d'itérations | 5 graines |
| 6.5 | Sensibilité / ablation | 1 instance par famille | $q_{max}$, $T_0$, recuit, adaptation | 3 graines |
| 6.6 | Rôle des opérateurs | résultats de 6.2 | opérateur de destruction | – |
| 6.7 | Coût des fenêtres de temps | 6 instances de Solomon | avec / sans fenêtres | 1 graine |
| 6.8 | Niveau B : CVRP | 5 instances X (Uchoa et al. 2017) | problème sans fenêtres | 3 graines |

**Familles de Solomon** : C = clients groupés (*clusters*), R = clients répartis aléatoirement, RC = mélange. Séries 1 = horizon court (fenêtres serrées, nombreuses petites tournées) ; séries 2 = horizon long (peu de longues tournées).

**Indicateur principal** : $gap = 100 \times \frac{\text{coût ALNS} - \text{coût BKS}}{\text{coût BKS}}$.

Tous les résultats sont enregistrés dans `results/*.csv`. Pour relancer une expérience, passer `FORCE_RERUN = True` ou supprimer le fichier CSV correspondant. Le temps d'exécution complet est d'environ 20 minutes sur un portable.

```python
FORCE_RERUN = False


def cached(name, fn):
    """Exécute fn() une seule fois et garde le résultat dans results/<name>.csv."""
    path = os.path.join(RES_DIR, f"{name}.csv")
    if os.path.exists(path) and not FORCE_RERUN:
        return pd.read_csv(path)
    df = fn()
    df.to_csv(path, index=False)
    return df


def relax_time_windows(data):
    """Même instance sans fenêtres de temps (CVRP)."""
    d = dict(data)
    d["e"] = np.zeros_like(data["e"])
    d["l"] = np.full_like(data["l"], 1e12)
    return d


def run_benchmark(names, collection, seeds, n_iter, rounding="dimacs", relax=False, **params):
    rows = []
    for name in names:
        inst, bks = load_instance(name, collection)
        data = build_data(inst, rounding)
        if relax:
            data = relax_time_windows(data)
        for seed in seeds:
            res = solve(data, n_iter=n_iter, seed=seed, **params)
            cost = check_solution(data, res["routes"])
            row = dict(instance=name, family=family(name), n=data["n"], seed=seed, n_iter=n_iter,
                       cost=cost, init_cost=res["init_cost"], bks=bks["cost"],
                       gap=100 * (cost - bks["cost"]) / bks["cost"],
                       init_gap=100 * (res["init_cost"] - bks["cost"]) / bks["cost"],
                       vehicles=len(res["routes"]), bks_vehicles=len(bks["routes"]),
                       time=res["time"])
            for k, op in enumerate(DESTROY_NAMES):
                row[f"best_{op}"] = res["best_destroy"][k]
                row[f"use_{op}"] = res["used_destroy"][k]
            rows.append(row)
    return pd.DataFrame(rows)
```

### 6.1. Validation par l'approche exacte (PLNE)
Le PLNE de la section 3 est construit à partir des matrices (contraintes (1) à (7)), puis résolu par HiGHS avec une limite de 60 s. On compare sur des sous-instances (les $n$ premiers clients d'instances de Solomon) l'optimum prouvé et la solution de l'ALNS.

```python
def sub_instance(data, n_sub):
    """Restreint une instance à ses n_sub premiers clients (même dépôt, même flotte)."""
    idx = np.arange(n_sub + 1)
    sub = {k: data[k][idx] for k in ("demand", "e", "l", "s")}
    sub["coords"] = data["coords"][idx]
    sub["dist"] = data["dist"][np.ix_(idx, idx)]
    sub.update(Q=data["Q"], K=data["K"], n=n_sub)
    return sub


def solve_milp(data, time_limit=60.0):
    """PLNE à deux indices (MTZ capacité + Big-M temporels) résolu par HiGHS (scipy.optimize.milp).
    Nœuds : 0 = dépôt de départ, 1..n clients, n+1 = dépôt d'arrivée (copie du dépôt)."""
    n = data["n"]
    t = data["dist"]
    e, l, s, q, Q, K = data["e"], data["l"], data["s"], data["demand"], data["Q"], data["K"]
    end = n + 1
    node = lambda i: 0 if i == end else i          # nœud physique
    E = lambda i: e[node(i)]
    Lw = lambda i: l[node(i)]
    S = lambda i: s[node(i)]
    T = lambda i, j: t[node(i), node(j)]

    arcs = [(i, j) for i in range(0, n + 1) for j in range(1, n + 2)
            if i != j and not (i == 0 and j == end)
            and E(i) + S(i) + T(i, j) <= Lw(j)]       # arcs temporellement impossibles retirés
    na = len(arcs)
    aid = {a: k for k, a in enumerate(arcs)}
    w0 = na                    # w_i, i = 0..n+1
    u0 = na + n + 2            # u_i, i = 1..n
    nv = u0 + n
    c = np.zeros(nv)
    for k, (i, j) in enumerate(arcs):
        c[k] = T(i, j)

    rows, lb, ub = [], [], []
    A = lil_matrix((4 * n + 2 + 2 * na, nv))
    r = 0
    for i in range(1, n + 1):                      # (1) chaque client quitté une fois
        for (a, b), k in aid.items():
            if a == i:
                A[r, k] = 1
        lb.append(1); ub.append(1); r += 1
    for h in range(1, n + 1):                      # (2) conservation du flot
        for (a, b), k in aid.items():
            if b == h:
                A[r, k] += 1
            if a == h:
                A[r, k] -= 1
        lb.append(0); ub.append(0); r += 1
    for (a, b), k in aid.items():                  # (3) au plus K véhicules
        if a == 0:
            A[r, k] = 1
    lb.append(0); ub.append(K); r += 1
    for (i, j), k in aid.items():
        if 1 <= i <= n and 1 <= j <= n:            # (4) MTZ charge : u_i + q_j <= u_j + Q(1-x_ij)
            A[r, u0 + i - 1] = 1
            A[r, u0 + j - 1] = -1
            A[r, k] = Q
            lb.append(-np.inf); ub.append(Q - q[j]); r += 1
        M = Lw(i) + S(i) + T(i, j) - E(j)          # (6) temps : w_i + s_i + t_ij <= w_j + M_ij(1-x_ij)
        if M > 0:
            A[r, w0 + i] = 1
            A[r, w0 + j] = -1
            A[r, k] = M
            lb.append(-np.inf); ub.append(M - S(i) - T(i, j)); r += 1
    A = A[:r].tocsr()

    lo = np.zeros(nv)
    hi = np.ones(nv)
    for i in range(n + 2):                         # (7) fenêtres de temps
        lo[w0 + i], hi[w0 + i] = E(i), Lw(i)
    for i in range(1, n + 1):                      # (5) capacité
        lo[u0 + i - 1], hi[u0 + i - 1] = q[i], Q
    integ = np.zeros(nv)
    integ[:na] = 1
    res = milp(c, constraints=LinearConstraint(A, lb, ub), integrality=integ,
               bounds=Bounds(lo, hi), options={"time_limit": time_limit, "disp": False})
    out = dict(status=res.status, message=res.message, cost=np.nan, bound=np.nan, routes=[])
    if res.x is not None:
        x = res.x[:na] > 0.5
        succ = {arcs[k][0]: arcs[k][1] for k in range(na) if x[k] and arcs[k][0] != 0}
        for k in range(na):
            if x[k] and arcs[k][0] == 0:
                route, cur = [], arcs[k][1]
                while cur != end:
                    route.append(cur)
                    cur = succ[cur]
                out["routes"].append(route)
        out["cost"] = res.fun
        out["bound"] = getattr(res, "mip_dual_bound", np.nan)
    return out


def exp_milp():
    rows = []
    for name in ["R101", "C101", "RC101", "R201", "C201", "RC201"]:
        data = build_data(load_instance(name, "solomon")[0])
        for n_sub in [10, 25, 50]:
            sub = sub_instance(data, n_sub)
            t = time.perf_counter()
            m = solve_milp(sub, time_limit=60)
            t_milp = time.perf_counter() - t
            a = solve(sub, n_iter=5000, seed=1)
            if m["routes"]:
                check_solution(sub, m["routes"])
            rows.append(dict(instance=name, n=n_sub,
                             milp_status="optimal" if m["status"] == 0 else "limite de temps",
                             milp_cost=m["cost"], milp_bound=m["bound"], milp_time=t_milp,
                             alns_cost=check_solution(sub, a["routes"]), alns_time=a["time"]))
    df = pd.DataFrame(rows)
    df["gap_alns_vs_milp"] = 100 * (df.alns_cost - df.milp_cost) / df.milp_cost
    df["gap_alns_vs_bound"] = 100 * (df.alns_cost - df.milp_bound) / df.milp_bound
    return df


df_milp = cached("milp_vs_alns", exp_milp)
df_milp.round(2)
```

### 6.3. Passage à l'échelle
**Gehring & Homberger** (200 à 1000 clients, même structure que Solomon) avec un budget fixe de 10 000 itérations, puis deux instances **générées aléatoirement** de 2000 et 3000 clients (sans BKS : on mesure l'amélioration par rapport à la solution initiale et le temps).

```python
GH = ["C1_2_1", "R1_2_1", "RC1_2_1", "C2_2_1", "R2_2_1", "RC2_2_1",
      "C1_4_1", "R1_4_1", "R1_6_1", "R1_8_1", "C1_10_1", "R1_10_1", "RC1_10_1"]
df_gh = cached("gh_scaling", lambda: run_benchmark(GH, "gh", seeds=[1], n_iter=10000))
df_gh_long = cached("gh_1000_long", lambda: run_benchmark(["R1_10_1"], "gh", seeds=[1], n_iter=30000))
pd.concat([df_gh, df_gh_long])[["instance", "n", "n_iter", "cost", "bks", "gap", "init_gap",
                                "vehicles", "bks_vehicles", "time"]].round(2)
```

```python
def generate_instance(n, layout="uniform", tw_width=100.0, capacity=200, seed=0, side=200.0):
    """Instance CVRPTW aléatoire (scénario « grande ville ») : dépôt central, clients uniformes
    ou groupés, fenêtres de largeur tw_width toutes atteignables depuis le dépôt."""
    rng = np.random.default_rng(seed)
    if layout == "uniform":
        xy = rng.uniform(0, side, (n, 2))
    else:
        centers = rng.uniform(0.1 * side, 0.9 * side, (max(3, n // 100), 2))
        xy = centers[rng.integers(len(centers), size=n)] + rng.normal(0, 0.04 * side, (n, 2))
        xy = np.clip(xy, 0, side)
    coords = np.vstack([[side / 2, side / 2], xy]).round(1)
    d0 = np.sqrt(((coords - coords[0]) ** 2).sum(1))
    horizon, service = 10 * side, 10
    e = rng.uniform(d0, horizon - d0 - service - tw_width)
    tw = np.column_stack([e, e + tw_width]).round()
    tw[0] = [0, horizon]
    demand = rng.integers(1, 30, n + 1)
    demand[0] = 0
    return dict(node_coord=coords, demand=demand, capacity=capacity, time_window=tw,
                service_time=service, vehicles=n // 5)


def exp_large():
    rows = []
    for n, layout in [(2000, "uniform"), (3000, "cluster")]:
        data = build_data(generate_instance(n, layout, seed=n))
        res = solve(data, n_iter=3000, seed=1)
        cost = check_solution(data, res["routes"])
        rows.append(dict(instance=f"gen-{n}-{layout}", n=n, n_iter=3000, cost=cost,
                         init_cost=res["init_cost"],
                         improvement=100 * (res["init_cost"] - cost) / res["init_cost"],
                         vehicles=len(res["routes"]), time=res["time"]))
    return pd.DataFrame(rows)


df_large = cached("generated_large", exp_large)
df_large.round(2)
```

```python
# Temps de calcul en fonction de n (budget fixe 10 000 itérations) : ajustement t = a * n^b
sz = pd.concat([df_sol[["n", "time"]], df_gh[["n", "time"]]]).groupby("n").time.mean()
b, log_a = np.polyfit(np.log(sz.index), np.log(sz.values), 1)
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
axes[0].loglog(sz.index, sz.values, "o", label="mesures (moyenne)")
nn = np.linspace(100, 1000, 50)
axes[0].loglog(nn, np.exp(log_a) * nn ** b, "--", label=f"ajustement  t ∝ n^{b:.2f}")
axes[0].set_xlabel("nombre de clients n")
axes[0].set_ylabel("temps (s) pour 10 000 itérations")
axes[0].legend()
axes[0].set_title("Temps de calcul vs taille")
g = pd.concat([df_sol.assign(src="Solomon"), df_gh.assign(src="GH")])
for src, grp in g.groupby("src"):
    m = grp.groupby("n").gap.mean()
    axes[1].plot(m.index, m.values, "o-", label=src)
axes[1].plot(df_gh_long.n, df_gh_long.gap, "r*", ms=12, label="R1_10_1, 30 000 itérations")
axes[1].set_xlabel("nombre de clients n")
axes[1].set_ylabel("gap moyen (%)")
axes[1].set_title("Qualité vs taille (budget fixe)")
axes[1].legend()
plt.tight_layout()
plt.show()
print(f"Exposant empirique : t ≈ {np.exp(log_a):.4f} × n^{b:.2f}")
```

### 6.5. Sensibilité aux paramètres et ablation
Approche « un facteur à la fois » autour du réglage par défaut ($q_{max} = 30$ pour $n = 100$, $T_0$ calibrée à 5 %, adaptation $\rho = 0.1$), sur une instance de chaque famille (C103, C203, R103, R203, RC103, RC203), 5000 itérations, 3 graines. Chaque variante est comparée au réglage par défaut par un **test de Wilcoxon apparié** (mêmes instances, mêmes graines).

```python
SENS_INST = ["C103", "C203", "R103", "R203", "RC103", "RC203"]
CONFIGS = {
    "défaut": {},
    "q_max = 10": dict(q_max=10),
    "q_max = 20": dict(q_max=20),
    "q_max = 50": dict(q_max=50),
    "T0 basse (0.5 %)": dict(start_worse=0.005),
    "T0 haute (20 %)": dict(start_worse=0.2),
    "sans recuit (descente)": dict(start_worse=1e-9),
    "sans adaptation (poids fixes)": dict(reaction=0.0),
}


def exp_sensitivity():
    dfs = []
    for label, params in CONFIGS.items():
        dfs.append(run_benchmark(SENS_INST, "solomon", seeds=[1, 2, 3], n_iter=5000, **params)
                   .assign(config=label))
    return pd.concat(dfs)


df_sens = cached("sensitivity", exp_sensitivity)
ref = df_sens[df_sens.config == "défaut"].set_index(["instance", "seed"]).gap
rows = []
for label in CONFIGS:
    g = df_sens[df_sens.config == label].set_index(["instance", "seed"]).gap.loc[ref.index]
    p = stats.wilcoxon(g, ref).pvalue if label != "défaut" and (g != ref).any() else np.nan
    rows.append(dict(config=label, gap_moyen=g.mean(), gap_max=g.max(),
                     temps=df_sens[df_sens.config == label].time.mean(), p_wilcoxon=p))
sens = pd.DataFrame(rows).set_index("config")
fig, ax = plt.subplots(figsize=(9, 4))
ax.barh(sens.index[::-1], sens.gap_moyen[::-1], color=["tab:red" if p < 0.05 else "tab:blue"
                                                       for p in sens.p_wilcoxon.fillna(1)[::-1]])
ax.set_xlabel("gap moyen (%)  –  rouge : différence significative avec le défaut (p < 0.05)")
plt.tight_layout()
plt.show()
pd.concat([sens.round(3), df_sens.pivot_table(index="config", columns="family", values="gap").round(2)], axis=1)
```
